## Exercise 1: Analytical exercises

In this exercise we derive the expressions for various derivatives of products of
vectors and matrices. Such derivatives are central to the optimization of various
cost functions. Although we will often use automatic differentiation in actual
calculations, having the analytical expressions is extremely helpful in case we
have simpler derivatives, as well as when we analyze various properties (like
second derivatives) of the chosen cost functions. Vectors are always written as
boldfaced lower-case letters and matrices as upper-case boldfaced letters.
Note that throughout we use the numerator (Jacobian) layout, in which the derivative of a
*scalar* with respect to a vector is a **row** vector — consistent with the example
$\partial(\boldsymbol{A}\boldsymbol{x})/\partial\boldsymbol{x}=\boldsymbol{A}$ in the hint below and with Faisal et al.; if you prefer the
column-gradient convention you will obtain the equivalent transposed results, e.g.
$(\boldsymbol{A}+\boldsymbol{A}^T)\boldsymbol{a}$ instead of $\boldsymbol{a}^T(\boldsymbol{A}+\boldsymbol{A}^T)$.

Show that

$$
\frac{\partial (\boldsymbol{a}^T\boldsymbol{x})}{\partial \boldsymbol{x}} = \boldsymbol{a}^T,
$$

and

$$
\frac{\partial (\boldsymbol{a}^T\boldsymbol{A}\boldsymbol{a})}{\partial \boldsymbol{a}} = \boldsymbol{a}^T\left(\boldsymbol{A}+\boldsymbol{A}^T\right),
$$

and

$$
\frac{\partial \left(\boldsymbol{x}-\boldsymbol{A}\boldsymbol{s}\right)^T\left(\boldsymbol{x}-\boldsymbol{A}\boldsymbol{s}\right)}{\partial \boldsymbol{s}} = -2\left(\boldsymbol{x}-\boldsymbol{A}\boldsymbol{s}\right)^T\boldsymbol{A},
$$

and finally find the second derivative of this function with respect to the vector
$\boldsymbol{s}$.

If we replace the vector $\boldsymbol{s}$ with the unknown parameters
$\boldsymbol{\theta}$ used to define the ordinary least squares method, we end up
with the equations that determine these parameters. The matrix $\boldsymbol{A}$ is
then the design matrix $\boldsymbol{X}$, and $\boldsymbol{x}$ has to be replaced
with the outputs $\boldsymbol{y}$. The second derivative of the mean squared error
is then proportional to the Hessian matrix
$\boldsymbol{H}=\boldsymbol{X}^T\boldsymbol{X}$.

**Hint:** it is always useful to write out with summation indices the various
quantities; take a look at the examples in the week 35 slides. As an example,
consider the function

$$
f(\boldsymbol{x}) = \boldsymbol{A}\boldsymbol{x},
$$

which reads for a specific component $f_i$ (defining $\boldsymbol{A}$ to be an
$n\times n$ matrix and $\boldsymbol{x}$ a vector of length $n$)

$$
f_i = \sum_{j=0}^{n-1} a_{ij} x_j,
$$

which leads to

$$
\frac{\partial f_i}{\partial x_j} = a_{ij},
$$

and written out in terms of the vector $\boldsymbol{x}$ we have

$$
\frac{\partial f(\boldsymbol{x})}{\partial \boldsymbol{x}} = \boldsymbol{A}.
$$

## Solution Excercise 1 ##

1. 
    to prove $ \frac{\partial (\boldsymbol{a}^T\boldsymbol{x})}{\partial \boldsymbol{x}} = \boldsymbol{a}^T $, we use the same method as showed above, as $ a^T \cdot x $ is perfectly linear that just leaves $ a^T $!

2. 
    to prove 
    $$
    \frac{\partial (\boldsymbol{a}^T\boldsymbol{A}\boldsymbol{a})}{\partial \boldsymbol{a}} = \boldsymbol{a}^T\left(\boldsymbol{A}+\boldsymbol{A}^T\right),
    $$
    we write the function in sumation notation: 
    $$
    (\boldsymbol{a}^T\boldsymbol{A}\boldsymbol{a}) = \sum_i \sum_j a_{j}A_{ij}a_{i}
    $$ 
    differentiating in respect to the k-th component:
    $$
    \frac{\partial f}{\partial a_k} = \sum_i \sum_j \left( \frac{\partial a_i}{\partial a_k} A_{ij} a_j + a_i A_{ij} \frac{\partial a_j}{\partial a_k} \right)
    $$
    That is just leaving a kronecker delta, as  $$\frac{\partial a_i}{\partial a_k} $$ is only differnt from 0, 
    when $$ k = i $$, that results in:
    $$\frac{\partial f}{\partial a_k} = \sum_j A_{kj} a_j + \sum_i a_i A_{ik}$$
    Converting back to vector notation (can just switch i to j, as it's just a dummy variable):
    $$\frac{\partial f}{\partial a_k} = \sum_j (A_{kj} + A_{jk}) a_j$$
    taking into account the mentioned convention, we see $a_j$ as a transposed vector, so we get the following result:
    $$\frac{\partial f}{\partial \boldsymbol{a}} = (\boldsymbol{A}\boldsymbol{a} + \boldsymbol{A}^T\boldsymbol{a})^T = \boldsymbol{a}^T\boldsymbol{A}^T + \boldsymbol{a}^T\boldsymbol{A} = \boldsymbol{a}^T (\boldsymbol{A} + \boldsymbol{A}^T)$$

3. 
    First we gonna apply the transpose rule : $ (AB)^T = B^T A^T $,, that leaves us with 
    $$f(\boldsymbol{s}) = (\boldsymbol{x}^T - \boldsymbol{s}^T\boldsymbol{A}^T) (\boldsymbol{x} - \boldsymbol{As})$$
    now we can multiply it out, like standard algebra: 
    $$f(\boldsymbol{s}) = \boldsymbol{x}^T\boldsymbol{x} - \boldsymbol{x}^T\boldsymbol{As} - \boldsymbol{s}^T\boldsymbol{A}^T\boldsymbol{x} + \boldsymbol{s}^T\boldsymbol{A}^T\boldsymbol{As}$$
    we see that both middle terms can be written in one, because the transposed of a scalar its just itself.
    $$f(\boldsymbol{s}) = \boldsymbol{x}^T\boldsymbol{x} - 2\boldsymbol{x}^T\boldsymbol{A}\boldsymbol{s} + \boldsymbol{s}^T(\boldsymbol{A}^T\boldsymbol{A})\boldsymbol{s}$$
    In this comprimised form we are able to use the rules from task 1 and task 2, so the first term just results in 0, as there is no dependency of s, the second term is just linear in s, so that one is just Task 1. Term 3 just applies Task 2, when we do this properly, we find the first derivative:

    $$\frac{\partial f}{\partial \boldsymbol{s}} = -2\boldsymbol{x}^T\boldsymbol{A} + 2\boldsymbol{s}^T\boldsymbol{A}^T\boldsymbol{A}$$

    We can factor out the two as well as the 2 and we find the given result. 

    To find the second derivative (the Hessian), we are deriving the first one another time. By doing so, the first term becomes zero and the second one is just linear with respect to s. The Hessian is just: 
    $$ \bold{H} = 2A^TA $$
 


## Exercise 2: Making your own data and exploring scikit-learn

We will generate our own dataset for a function $y(x)$ where $x\in[0,1]$ and
defined by random numbers computed with the uniform distribution. The function $y$
is a quadratic polynomial in $x$ with added stochastic noise according to the
normal distribution $\mathcal{N}(0,1)$. The following simple Python instructions
define our $x$ and $y$ values (with 100 data points).

```python
import numpy as np

rng = np.random.default_rng(2026)
x = rng.random((100, 1))
y = 2.0 + 5.0 * x * x + 0.1 * rng.standard_normal((100, 1))
```

1. Write your own code (following the examples in the book, Chapter 3, and the

3. Using scikit-learn, compute also the mean squared error, a risk metric
   corresponding to the expected value of the squared (quadratic) error

$$
MSE(\boldsymbol{y},\tilde{\boldsymbol{y}}) = \frac{1}{n}\sum_{i=0}^{n-1}\left(y_i-\tilde{y}_i\right)^2,
$$

   and the $R^2$ score function. If $\tilde{y}_i$ is the predicted value of the
   $i$-th sample and $y_i$ is the corresponding true value, then the $R^2$ score is

$$
R^2(\boldsymbol{y},\tilde{\boldsymbol{y}}) = 1 - \frac{\sum_{i=0}^{n-1}(y_i-\tilde{y}_i)^2}{\sum_{i=0}^{n-1}(y_i-\bar{y})^2},
$$

   where the mean value of $\boldsymbol{y}$ is

$$
\bar{y} = \frac{1}{n}\sum_{i=0}^{n-1} y_i.
$$

   You can use the functionality included in scikit-learn, or your own functions
   from the Tuesday notebook. Discuss the meaning of these results. Try also to
   vary the coefficient in front of the added stochastic noise term and discuss
   the quality of the fits.

```python
#exercise 1
#creating the dataset
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

rng = np.random.default_rng(2026)
x = rng.random((100, 1))


for noise in np.arange(0.0, 5.0, 0.5):
    y = 2.0 + 5.0 * x * x + noise * rng.standard_normal((100, 1)) 
    #Build the degree 2 design matrix
    X = np.column_stack((np.ones_like(x), x, x**2))

    #Doing SVD
    U, S, VT = np.linalg.svd(X, full_matrices=False) # full_matrices=False to get the reduced SVD, dropping redundant columns 
    #S is now a 1D array of singular values, we need to convert it into a diagonal matrix
    S_matrix = np.diag(1/S)

    theta = VT.T @ S_matrix @ U.T @ y

    #doing the same with skykit-learn:

    #Building the design matrix 
    poly = PolynomialFeatures(degree=2, include_bias=False) # set bias to False to avoid adding an extra column of ones, as the mode is already using a centered matrix 
    X_poly = poly.fit_transform(x)

    #fit this model using linear regression
    model = LinearRegression()
    model.fit(X_poly, y)

    #computing the r2 value as well as the mean squared error 
    mse = mean_squared_error(y, model.predict(X_poly))
    r2 = r2_score(y, model.predict(X_poly))

    # Output formatted as a list for the current noise coefficient
    current_output = [
        f"Noise factor: {noise:.1f}",
        'SVD fittet coefficients: Intercept: {:.4f}, Linear term: {:.4f}, Quadratic term: {:.4f}'.format(theta[0, 0], theta[1, 0], theta[2, 0]),
        f"SKLearn Intercept (theta_0): {model.intercept_[0]:.2f}",
        f"SKLearn Coefficients (theta_1, theta_2): {model.coef_[0, 0]:.2f}, {model.coef_[0, 1]:.2f}",
        f"Mean Squared Error: {mse:.4f}",
        f"R^2 Score: {r2:.4f}"
    ]
    print(current_output)
```

## Dicussing the MSE ans the R2:
The R2 describes how good the model actually fits the data, the closer the value is to 1, the better fit it is. As the given value is approximately 0.993, die second degree polynomial is a good fit for our data. 

The MSE represents the variance of the residuals, as we scale the noise by 0.1 and it is given by a Gaussion Distribution, the MSE should (as it does) around 0.01. The model is capturing the curve perfectly, but fails to predict the noise. 
By raising the noise coefficient, the MSE increases rapidly and the model fails to predict the data as the variance rises.

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
import matplotlib.pyplot as plt


rng = np.random.default_rng(2026)

n = 100
x = np.linspace(-3, 3, n).reshape(-1, 1)
y = np.exp(-x**2) + 1.5 * np.exp(-(x - 2)**2) + rng.normal(0, 0.1, x.shape)

train_mse = []
test_mse = []
for i in range(1,16):
    x_train, x_test, y_train, y_test = train_test_split(
        x, y, test_size=0.2, random_state=42
    )

    poly = PolynomialFeatures(degree=i, include_bias=False)
    x_poly_train = poly.fit_transform(x_train)
    x_poly_test = poly.transform(x_test)

    model = LinearRegression()
    model.fit(x_poly_train, y_train)

    mse_train = mean_squared_error(y_train, model.predict(x_poly_train))
    r2_train = r2_score(y_train, model.predict(x_poly_train))

    mse_test = mean_squared_error(y_test, model.predict(x_poly_test))
    r2_test = r2_score(y_test, model.predict(x_poly_test))

    train_mse.append(mse_train)
    test_mse.append(mse_test)

print("Train MSE for degrees 1 to 15:", train_mse)
print("Test MSE for degrees 1 to 15:", test_mse)

# Find and print the minimal Test MSE and its degree
min_test_mse = min(test_mse)
best_degree = test_mse.index(min_test_mse) + 1
print(f"\nMinimal Test MSE: {min_test_mse:.4f} (at degree {best_degree})")

plt.plot(range(1,16), train_mse, label='Train MSE')
plt.plot(range(1,16), test_mse, label='Test MSE')
plt.xlabel('Polynomial Degree')
plt.ylabel('MSE')
plt.legend()
plt.show()
```

## Discussion of the results 
As we have seen in class overfitting often results in some sort of U-curve, this isn't happening here. With 100 data point we are not actually overfitting, we are having 80 data points pinning down the curve. Also we are using linspace, our data is perfectly and even distributed, almost every test point in sandwhiched by two training points. Thats why we cant see the known u-curve, by using a higher degree polynomial we can actually see it for the test mse.
